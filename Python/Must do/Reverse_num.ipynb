# Write a program to reverse an integer in Python.
# Example:
# ○ Input: 1234
# ○ Output: 4321

```python
num = int(input("Input:"))

reverse_num = 0 

while num > 0:
    reverse_num = (reverse_num*10)+num%10
    num  = num //10
print("Output:",reverse_num)    
    
```

```python
# Using Def function
def reverse_number(num):
    reverse = 0
    while num > 0:
        reverse = (reverse * 10)+num%10
        num = num // 10
    return reverse


num = int(input("Enter number : "))
print("output : ",reverse_number(num))
```
