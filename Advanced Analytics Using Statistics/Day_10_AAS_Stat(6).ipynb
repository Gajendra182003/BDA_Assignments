```python
import polars as pl

# 1. Create or read a DataFrame
df = pl.DataFrame(
    {
        "id": [1, 2, 3, 4],
        "category": ["A", "B", "A", "B"],
        "value": [10.5, 20.0, 15.0, 30.2],
    }
)

# 2. Filter and aggregate (Expression Syntax)
result = (
    df.filter(pl.col("category") == "A")
    .group_by("category")
    .agg(
        pl.col("value").sum().alias("total_value"),
        pl.col("id").count().alias("count")
    )
)

print(result)
```
