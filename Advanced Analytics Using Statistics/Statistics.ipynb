# Types of data

#### 1. Qualitative data: Mostly in strings or words
#### 2. Quantitative data: Mostly in numbers

1. Quantitative Data 
- Continuous Data:
  - Continuous data is the one where possibility of decimal point cannot be ruled out.
  - Will always be the result of measurememt (for measurement we need instrument)
  - It will always be expressed in "mlt"
- Discrete Data:
  - Descrete data will never have decimal point.
  - It will always be a result of counting.
  - If n*p and n*(1-p)>10, then descrete data can be considered as continuous. (n -> sample size, p -> proportion)
  - For large sample size, descrete and continuous can be considered to be same.
- Any rating or ranking even if it is a number will be considered as categorical data.

Specifically in regression:
Type of scale:
1. Nominal: Got multiple category, but we may choose one or more.(there is no logical order) (Categorical data)
2. Ordinal: Ranked or rating. (It has logical order)(Comparision) (Categorical Data)
3. Interval: Zero does not mean absence of measurement, so mathematical calculations will be restricted. You compare the values but cannot take the ratio of the values

1. Mean: avg
- Mean should be used only when there are no extreme valeus in the dataset.
- Mean should be use categoricald only when the data is normally/symmetrically distributed.

- It is the only parameter which can be caluclated for qualitative data and it can have more than 1 value.

```python
import pandas as pd
import numpy as np

np.random.seed(42)

data = np.random.normal(loc=50, scale=15, size=1990)

outliers = np.array([5000, 5200, 5500, 5800, 6000, 6200, 6500, 6800, 7000, 7200])

data_list = np.concatenate([data, outliers])

df = pd.DataFrame(data_list, columns=['Values'])
q1 = df['Values'].quantile(0.25)
q3 = df['Values'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

print(f"Q1: {q1}")
print(f"Q3: {q3}")
print(f"IQR: {iqr}")
print(f"Upper Outlier Threshold: {upper_fence}")
print(f"Lower Outlier Threshold: {lower_fence}")
```

# Shape -> skewness and kurtosis
1. Skewness:
- In graph, if peak is on the left side
    - distribution is skewed to right
    - positively skewed
    - skewness > 0
- In graph, if peak is in the center
    - distribution is symmetric
    - skewness = 0
- In graph, if the peak is on the right side
    - distribution is skewed to left
    - negatively skewed
    - skewness < 0

# Probability Distribution
1. Distribution for discrete data.
- Binomial Distribution (Used when there are only two possible occurrences and probability of occurrences is known)
  Mean no of occurrences equal to n*p
  e.g. Chances of rain on any given day in June in Mumbai is 0.7
  If I am in Mumbai for 15 days, what is the probability that
  a. It rains for exactly 10 days.
  b. It rains for maximum 10 days.
  c. It rains for minimum 10 days.
- Poisson Distribution (Used when we are more concerened about frequency of an event -> how many times it occur)
  
2. Distribution for continuous data
- Exponential Distribution
  It is used to model the time gap between two successive events.
- Normal Distribution
  It is a normal bell shape curve (to define it we need mean and std dev)
  Standard Normal Curve (mean=0 and std_dev=1)
- Students t distribution
  It is used when the population standard deviation is not known, what we know is the sample standard deviation
  Necessarily used when the sample size is less than equal to 30
  For small samples, normal distribution cannot be used
  e.g. 200 incidents happening in a year so mean time between failure (MTBF) is (356*24)/200 unit is hours/incident
  1 incident-> 1/(356*24)/200 is rate of failure
