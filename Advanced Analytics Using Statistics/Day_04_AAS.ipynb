# **MERGING of DATAFRAME**

```python
pune = pd.read_excel('CDAC_DataBook.xlsx', sheet_name='Pune')
```

```python
pune
```

```python
mumbai = pd.read_excel('CDAC_DataBook.xlsx',sheet_name = 'Mumbai')
mumbai
```

```python
pd.merge(pune,mumbai, on='Subject')
```

```python
pd.merge(pune,mumbai, on='Subject', how= 'inner')
```

```python
pd.merge(pune,mumbai, on=['Subject','Grade'],how='inner')
```

```python
df2_sort_values.info()
```

```python
pd.merge(pune,mumbai,on=['Subject','Grade'])
```

```python
pd.merge(pune,mumbai,on='Subject',how='outer')
```

```python
print("Pune Column")
print(pune)

print('==============================')
print("Mumbai Column")
print(mumbai)
```

```python
print(pd.merge(pune,mumbai,on='Subject',how='outer'))
```

```python
print('Left\n')
print(pd.merge(pune,mumbai,on='Subject',how='left'))
print('=================================================')
print('Right\n')
print(pd.merge(pune,mumbai,on='Subject',how='right'))
```
