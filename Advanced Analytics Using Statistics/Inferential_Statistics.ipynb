# Inferential Statistics

#### Sample statistic changes everytime we draw a sample, but population parameter remains constant
## Central Limit Theorem (CLT)
- If multiple samples are collected from a population and are plotted on a graph. then irrespective of population distribution the distributions of the sample mean will always follow normal distribution. (no of samples should be more than 30)
- The mean of all the sample means is equal to population mean.
- The standard deviation of the sample mean is equal to standard deviation of population divided by square root of sample size.

# Hypothesis Testing
## Step 1: Establish the null (Ho) and alternative (Ha) hypothesis.
### 1. Null Hypothesis
It's made in such a way that if it is found to be true then:
   - No action is required.
   - No decision is required.
   - No changes are required.
Null Hypothesis : Prover and Defender Approach. (Most of the time we will use defender approach)
e.g. Min salary is 40K
- If sal>=40K Null Hypothesis
- If sal<40K Alternate Hypothesis
e.g. Max delivery time is 20min
- If delivery_time <=20 Null Hypothesis
- If delivery_time >20 Alternate Hypothesis
e.g. Marks required to qualify is 60:
- If marks>=60 Null Hypothesis
- If marks<60 Alternate Hypothesis

## Step 2: Establish the confidence_level and the level of significance
- Level of Significance = 100-confidence_level (in percentage)
- Level of Significance = 1-confidence_level (in fraction)

## Step 3: Choose the appropriate statistical test.
- p-value: Assuming the null hypothesis to be true p-value gives the probability of finding a value as extreme as we have found/observing.
## Step 4: Calculate the value of the test statistic and p-value

## Step 5: Compare the p-value(from step 4) against the level of significance(from step 2)
> ### IF THE P-VALUE IS LESS THAN THE LEVEL OF SIGNIFICANCE THEN WE REJECT THE NULL HYPOTHESIS.
> ### IF THE P-VALUE IS MORE THAN THE LEVEL OF SIGNIFICANCE THEN WE DO NOT REJECT THE NULL HYPOTHESIS.
> ### Decisions are only taken on the null hypothesis

## Type 1 Error: It is committed when we reject a null hypothesis which is actually true.
## Type 2 Error: It is committed when we do not reject a null hypothesis which is actually false.
                   Decision
    Ho    |  Reject   |   Do not reject
    ------+-----------+----------------
    True  |  Type I   |     Correct
          |           |
    False |  Correct  |     Type-II
