+ 2-proportion test \
This test is used to compare the difference of two proportions calculated from two samples against a claim value and establish whether their difference is statistically significant or not

# Chi Square test : for goodness of fit

+ Used when there are multiple categories. The counts obtained for each category would be compared aginst the claimed pattern to check whether the counts follow the pattern or not. The standard Ho is that the counts follow the pattern

```python
# Claimed pattern for the qualifications
# Btech - 0.4, Mtech - 0.1, MCA - 0.3, MSc - 0.2
# Btech - 50, Mtech - 20, MCA - 45, MSc - 25
# Whether counts obtained from sample follow the pattern or not.
```

```python
counts = np.array([50,20,45,25])
```

```python
pattern = np.array([0.4,0.1,0.3,0.2]) 
```

```python
exp_counts = sum(counts) * pattern
```

```python
exp_counts
```

```python
from scipy.stats import chisquare
```

```python
scipy.stats.chisquare(counts, f_exp = exp_counts)
```

```python
counts = np.array([70,10,35,45])
from scipy.stats import chisquare
exp_counts = sum(counts)*pattern
scipy.stats.chisquare(counts, f_exp = exp_counts)
```
