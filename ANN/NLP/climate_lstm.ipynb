```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error

from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
```

```python
df=pd.read_csv("DailyDelhiClimateTrain.csv")
df.head()
```

```python
temp=df[["meantemp"]]
temp.head()
```

```python
plt.plot(temp)
plt.show()
```

```python
print(temp.shape)
```

```python
from statsmodels.tsa.stattools import acf
```

```python
acf_values=acf(temp)
print(acf_values)
```

```python
scalar=MinMaxScaler()
scaled_temp=scalar.fit_transform(temp)
```

```python
scaled_temp
```

```python
look_back=30
```

```python
X=[]
y=[]
for i in range(len(scaled_temp)-look_back):
    X.append(scaled_temp[i:i+look_back])
    y.append(scaled_temp[i+look_back])
X=np.array(X)
y=np.array(y)
```

```python
X.shape
```

```python
split=int(len(X)*0.8)

X_train= X[:split]
X_test=X[split:]

y_train=y[:split]
y_test=y[split:]
```

```python
X_train.shape
X_test.shape
```

```python
#from sklearn.model_selection import train_test_split
#X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,shuffle=False)
```

```python
X_train.shape
```

```python
model=Sequential([
LSTM(units=128,return_sequences=True,input_shape=(look_back,1)),
Dropout(0.2),

LSTM(64,return_sequences=True),
Dropout(0.2),

LSTM(32),
Dropout(0.2),

Dense(1)
])
```

```python
model.compile(optimizer='adam', loss='mean_squared_error')
```

```python
model.summary()
```

```python
early_stop=keras.callbacks.EarlyStopping(monitor='val_loss',patience=10,restore_best_weights=True)
```

```python
history=model.fit(X_train,y_train,epochs=100,batch_size=32,validation_split=0.2,
                  callbacks=[early_stop])
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('model loss')
plt.ylabel('loss')
plt.xlabel('epoch')
plt.legend(['train', 'validation'], loc='upper left')
```
