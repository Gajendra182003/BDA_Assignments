```python
import nltk
from nltk.stem import PorterStemmer
```

```python
stem1=PorterStemmer()
```

```python
words=['run','running','ran','runs','easily','fairly','happy','affection','caring']
```

```python
for word in words:
    print(word+'---->'+stem1.stem(word))
```

LEMMATISATION

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lem1=WordNetLemmatizer()
```

```python
words=['run','running','ran','runs','easily','fairly','happy','affection','caring','cars','happiness']
```

```python
for word in words:
    print(word+'---->'+lem1.lemmatize(word))
```

```python
lem1.lemmatize('better')
```

```python
lem1.lemmatize('better',pos='a')
```
